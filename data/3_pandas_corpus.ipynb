## 5. Filtrer des lignes

`grc["pos"] == "V"` répond `True` ou `False` pour chaque ligne. Mis entre crochets, `grc[...]` ne garde que les lignes où la réponse est `True`.

## 6. Ajouter une colonne

On crée une colonne comme on remplit une variable : `tableau["nouvelle_colonne"] = ...`.
`.str` donne accès aux opérations sur le texte, appliquées à toute la colonne d'un coup.

````python
grc["longueur"] = grc["form"].str.len()      # nombre de lettres de chaque forme
grc.head()
````

Un `@` dans `pos` signale une **forme composée** : un seul mot écrit qui en réunit plusieurs, par exemple « article + adjectif » (`DET@A`).

````python
grc["composee"] = grc["pos"].str.contains("@")   # True si la catégorie contient un @
grc[grc["composee"]].head()
````

````python
print(grc["composee"].sum())    # True compte pour 1 et False pour 0 : sum() donne le nombre de True
````

## 7. Trier : `sort_values`

````python
grc.sort_values("longueur", ascending=False).head()    # ascending=False : du plus grand au plus petit
````

## 8. Calculer par groupe : `groupby`

« Pour chaque catégorie, quelle est la longueur moyenne des formes ? » En pandas : on **groupe** par `pos`, on prend la colonne `longueur`, on fait la **moyenne**.

````python
grc.groupby("pos")["longueur"].mean()
````

````python
grc.groupby("pos")["longueur"].mean().sort_values(ascending=False).head()   # les 5 catégories aux formes les plus longues
````

## 9. Lire tous les fichiers et les réunir : `concat`

Il y a 12 fichiers : 4 langues × 3 parties (`train`, `test_in-domain`, `test_out-of-domain`). On les lit un par un dans une boucle, on ajoute à chacun deux colonnes qui disent d'où il vient, puis `pd.concat` les empile en un seul tableau.

````python
tableaux = []

for partie in ["train", "test_in-domain", "test_out-of-domain"]:
    dossier = os.path.join(DATA, "corpus", partie)
    for nom_fichier in sorted(os.listdir(dossier)):
        if nom_fichier.endswith(".tsv"):                 # on ignore tout autre fichier
            tableau = pd.read_csv(os.path.join(dossier, nom_fichier), sep="\t")
            tableau["langue"] = nom_fichier[:3]          # "GRC_train.tsv" → "GRC"
            tableau["partie"] = partie
            tableaux.append(tableau)

corpus = pd.concat(tableaux, ignore_index=True)          # ignore_index : renuméroter les lignes de 0 à la fin
corpus.shape
````

````python
corpus.sample(5)    # 5 lignes au hasard, de n'importe quelle langue
````

`pd.crosstab` fait un **tableau croisé** : ici, le nombre de mots pour chaque langue et chaque partie.

````python
pd.crosstab(corpus["langue"], corpus["partie"])
````

## 11. Les mots jamais vus : `isin`

Les modèles sont évalués sur les parties `test`. Question : parmi les mots du test, combien apparaissent déjà dans le `train` ?
`isin` répond `True` si la valeur se trouve dans une autre liste.

````python
train_grc = corpus[(corpus["langue"] == "GRC") & (corpus["partie"] == "train")]
test_grc = corpus[(corpus["langue"] == "GRC") & (corpus["partie"] == "test_in-domain")]

deja_vus = test_grc["form"].isin(train_grc["form"])
print(round(deja_vus.mean(), 2))     # proportion de mots du test déjà présents dans le train
````

Même calcul pour chaque langue et chaque test, avec deux boucles :

````python
for langue in ["GRC", "HYE", "KAT", "SYC"]:
    train = corpus[(corpus["langue"] == langue) & (corpus["partie"] == "train")]
    for partie in ["test_in-domain", "test_out-of-domain"]:
        test = corpus[(corpus["langue"] == langue) & (corpus["partie"] == partie)]
        proportion = test["form"].isin(train["form"]).mean()
        print(langue, partie, round(proportion, 2))
````

`in-domain` : même texte que le train. `out-of-domain` : un autre auteur, un autre genre. Hors domaine, il y a moins de mots déjà vus (en grec : 73 % contre 51 %), donc la tâche est plus difficile.

## 12. Associer deux tableaux : `merge`

Une méthode très simple pour annoter le test : pour chaque mot, reprendre la catégorie qu'il avait dans le train. C'est un **dictionnaire**, et il nous servira de point de comparaison avec les LLM.

1. Le dictionnaire : une ligne par forme (`drop_duplicates` garde la première rencontrée).

````python
dico = train_grc[["form", "pos"]].drop_duplicates("form")
dico = dico.rename(columns={"pos": "pos_dico"})
dico.head()
````

2. `merge` ajoute au test la colonne `pos_dico`, en associant les lignes qui ont la même `form`. `how="left"` garde **tous** les mots du test, même ceux qui ne sont pas dans le dictionnaire.

````python
resultat = test_grc.merge(dico, on="form", how="left")
resultat[["form", "pos", "pos_dico"]].head(10)
````

`NaN` (*Not a Number*) signifie « valeur manquante » : le mot n'était pas dans le train.

3. Score : la proportion de lignes où les deux catégories sont égales.

````python
print(round((resultat["pos"] == resultat["pos_dico"]).mean(), 2))
````

Le même chiffre que la partie 11 : le dictionnaire trouve la bonne catégorie de presque tous les mots déjà vus, et ne peut rien dire des autres. Toute la difficulté est dans les mots jamais vus ; c'est là qu'un LLM peut faire mieux.

## Récapitulatif

| Je veux… | pandas |
|---|---|
| lire un TSV | `pd.read_csv(chemin, sep="\t")` |
| voir le début, la fin | `df.head()`, `df.tail()` |
| connaître la taille, les colonnes | `df.shape`, `df.columns`, `df.info()` |
| un résumé | `df.describe()` |
| une colonne, plusieurs colonnes | `df["col"]`, `df[["col1", "col2"]]` |
| compter les valeurs | `df["col"].value_counts()`, `.nunique()`, `.unique()` |
| filtrer des lignes | `df[df["col"] == valeur]`, `df[(condition1) & (condition2)]` |
| ajouter une colonne | `df["nouvelle"] = ...`, `.str.len()`, `.str.contains("@")` |
| trier | `df.sort_values("col", ascending=False)` |
| calculer par groupe | `df.groupby("col")["autre"].mean()` |
| réunir plusieurs tableaux | `pd.concat(liste_de_tableaux, ignore_index=True)` |
| tableau croisé | `pd.crosstab(df["col1"], df["col2"])` |
| tester l'appartenance | `df["col"].isin(autre_liste)` |
| associer deux tableaux | `df.merge(autre, on="col", how="left")` |
| renommer, dédoublonner | `df.rename(columns={"a": "b"})`, `df.drop_duplicates("col")` |
| enregistrer | `df.to_csv("fichier.csv", index=False)` |
